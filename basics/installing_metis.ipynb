# Installing Metis

Metis is implemented in Python and its dependencies are manged with
[uv](https://docs.astral.sh/uv/). While you can install Metis without
uv, this is **not recommended**.

## Installation with uv

Before you begin install uv and git as system packages on your operating
system. Then,

1.  Clone the Metis repository via

    ``` bash
    git clone ...
    ```

2.  Change directory

    ``` bash
    cd ...
    ```

3.  Use uv to install the correct version of Python and all of the
    required dependencies

    ``` bash
    uv sync
    ```

    > **Tip**
    >
    > If you would like to contribue to the development of Metis (or
    > just run the test suite), use
    >
    > ``` bash
    >  uv sync --dev
    > ```
    >
    > and then
    >
    > ``` bash
    > uv run pytest -v
    > ```
    >
    > to run the full test suite.
