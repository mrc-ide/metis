# Training in METIS

The entrypoint to a training job in Metis is `train.py`. Rather than
importing training code or writing a script, in Metis we simply run
`train.py` in the terminal or a notebook with the appropriate config
files supplied. Given the [data and config file we prepared
earlier](preparing-data.qmd), getting started with training is a matter
of overriding the default data config.

> **Tip**
>
> To follow along, either run the commands from the repository root in a
> terminal or download this page as a notebook and run through the
> cells. In a terminal, omit the notebook-only `%%bash` lines.

## Overriding the default training config

First we ensure the repository root is our current directory since this
is where `train.py` is located.

```python
import os
from IPython.display import Image

if os.path.basename(os.getcwd()) == "examples":
    os.chdir("..")
print("cwd:", os.getcwd())
```

Now we’ll override the default training config to use our `sirds`
dataset and in order to check that our model fit the data well we will
generate some samples via the sampling arguments:

-   `sampling.generate_samples=True` indicates we’d like the surrogate
    to generate samples once we’re done training
-   `sampling.samples_name` specifies where we would like the samples
    saved
-   `sample.subsample` used to generate samples for only a subset of the
    validation set

Now go ahead and run this:

```python
%%bash
uv run train.py \
  --config-name default \
  dataset=sirds \
  sampling.generate_samples=True \
  sampling.samples_name=examples/mlp_plot.png \
  sampling.subsample=5
```

If you’re in a notebook run this cell to see the model’s plots,
otherwise navigate to `examples/mlp_plot.png` and take a look there. The
orange lines are realisations from the SIRDS simulator for a particular
parameter group and the blue line is the prediction from our MLP
surrogate.

```python
Image(filename="examples/mlp_plot.png")
```

## Training a transformer

Next we’ll try a transformer. All we need to do is override the model
config. Be aware that this may take a bit longer to train:

```python
%%bash
uv run train.py \
  --config-name default \
  dataset=sirds \
  sampling.generate_samples=True \
  sampling.samples_name=examples/transformer_plot.png \
  sampling.subsample=5 \
  model=transformer_default
```

As before we can view the model’s predictions:

```python
Image(filename="examples/transformer_plot.png")
```

## Training a flow matching model

Again all we need to do is swap out the model config. Since this model
is stochastic we generate multiple samples per parameter group. Solid
blue is the mean and light blue are individual samples.

```python
%%bash
uv run train.py \
  --config-name default \
  dataset=sirds \
  sampling.generate_samples=True \
  sampling.samples_name=examples/flow_matching_plot.png \
  sampling.subsample=5 \
  model=flow_matching_default
```

```python
Image(filename="examples/flow_matching_plot.png")
```
