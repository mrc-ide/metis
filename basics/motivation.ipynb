# Motivating surrogate modelling

## Why surrogate modelling?

Many scientific disciplines are increasingly reliant on computational
simulators which reproduce complex behaviours. However, these simulators
are often slow and expensive to run, rendering tasks like scenario
analysis, forecasting and parameter inference, difficult or otherwise
impractical. For example, its been reported that a single forward
simulation over a large COVID-19 Agent-Based Model like JUNE can take
several days (Chopra et al. 2023). Surrogate models serve as
cost-effective approximations of these systems. Whilst surrogates have
been implemented with techniques like nonparametric statistical methods
and splines, modern neural network approaches offer significantly
greater flexibility and are often more performant.

> **Tip**
>
> For an overview of the surrogate modelling process and a reporting
> standard for surrogates, see SMRS (Semenova et al. 2025).

## Introducing METIS

METIS is a set of configurable recipes for training neural surrogates of
time series simulators. METIS implements several neural network
architectures and algorithms and is designed so that practictioners can
easily train and sweep over different combinations purely through simple
config files and comandline overrrides. Using METIS to develop
surrogates for existing simulators simply requires preparing data in the
correct format and supplying appropriate config files.
