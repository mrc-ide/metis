# METIS

Configurable neural network surrogates for stochastic timeseries
simulators

METIS is a collection of configurable “recipes” for training neural
network surrogate (emulator) models that stand in for expensive
stochastic simulators. Instead of running a slow simulator every time
you need a sample, you train a model once and then generate realisations
cheaply.

The codebase favours simple, single-file implementations wired together
with [Hydra](https://hydra.cc) configuration, so that you can swap
models, datasets and training settings by editing YAML rather than code.

### Overview of METIS features

-   Implementations of MLPs, Autoregressive Transformers, Flow Matching
    and Normalising Flows.
-   Support for static parameter conditioning and sequence conditioning.
-   Parquet-based dataloading pipeline.
-   Weights and Biases integration for hyperparameter sweeping and
    experiment tracking.
-   Visualisation utilities.
